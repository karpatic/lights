## Current Status

- The V2 controller is not fabrication-ready and not sale-ready.
- The ESP32-S3 module, antenna keepout, enclosure, labeling, firmware behavior, and final product use case all affect compliance work.
- Before selling, marketing, importing, or distributing assembled hardware, verify current FCC/ISED/CE/UKCA, Bluetooth SIG, product safety, labeling, and battery-related obligations with qualified professionals and current official requirements.
- A bare LiPo is not supported by the current V2 hardware design, so battery compliance should not be implied by any archived note.
