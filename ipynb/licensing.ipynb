# Bare-minimum compliance and cheapest-to-production game plan

> **Research summary, not legal advice.** Verified against current primary sources on 2026-08-04. Requirements can change, and the exact module grant, firmware, enclosure, supply, cables, sales channel, and intended use control the result.

## The product this plan covers

The cheapest first U.S. SKU should be:

- the current custom controller using **ESP32-S3-MINI-1-N8** with its integrated antenna;
- one plastic, nonconductive enclosure family;
- regulated **5 V input only**;
- no internal or bundled lithium battery;
- no bundled AC adapter or LED rope for the first SKU;
- no additional radio, display, motor, microphone, camera, or cloud account;
- one frozen PCB, antenna location, firmware radio configuration, cable set, and current rating.

The hardware remains design-in-progress and is not fabrication-ready. The firmware still needs the ESP32-S3 target, GPIO18 alignment, and an enforced current ceiling before a compliance sample can represent production. See the [current V2 design record](./hardware/compact-esp32-s3-led-controller-v2/DESIGN.md).

## Bottom line

**There is normally no new FCC certificate or license for the finished controller.** The low-cost FCC path is:

1. rely on the radio module's existing FCC certification, within every condition of its grant;
2. measure the finished controller against Part 15B Class B limits;
3. issue our own Supplier's Declaration of Conformity (SDoC);
4. retain the test and production records;
5. supply the required label and user information.

The selected module's published grant identifies **FCC ID `2AC7Z-ESPS3MINI1`** and single-modular approval. It also says the final integrator must assess the composite product and that portable or different-antenna configurations can require separate approval. The exact current host-integration exhibit and RF-exposure conditions must therefore be closed before release.

## 1. Required by U.S. law before marketing

### Preserve the module authorization

- Use the exact certified module and onboard antenna.
- Follow Espressif's land pattern, antenna keepout, supply, grounding, and host-integration instructions.
- Keep firmware within the approved channels, power, bandwidth, modulation, and regional settings.
- Do not add another transmitter or change the antenna without a new review.
- Confirm that normal and reasonably foreseeable use satisfies the grant's RF-exposure conditions.

If the integration leaves the grant, new transmitter work can require an accredited RF laboratory, a Telecommunication Certification Body, and possibly cooperation from Espressif. Avoiding that is the largest FCC cost-control decision.

### Establish host Part 15B Class B compliance

The custom PCB, clocks, regulator, GPIO/data switching, cables, and connected representative load make the finished consumer controller a Class B digital device. The certified module does **not** cover this unintended noise.

SDoC is self-authorization, but it is not measurement-free. A production-representative worst-case unit must have defensible conducted and radiated results using the applicable FCC/ANSI procedures. USB power does not automatically remove conducted testing: when the controller can operate through an AC-powered USB supply, §15.107(d) ordinarily requires that indirect AC-powered configuration to be measured.

> **There is no automatic “safe list.”** A 5 V USB input, plastic enclosure, PWM below 150 kHz, low bus speeds, short traces, no charger, and a module FCC ID can reduce risk; none of them independently authorizes the host or proves that it meets the measured limits.

For SDoC:

- no FCC application or filing fee;
- no new FCC ID;
- no FCC-issued Part 15B certificate;
- no TCB or accredited laboratory legally required;
- valid measurements, a proper measurement facility, and complete retained records are required;
- SDoC cannot be used when §2.906(d)'s FCC Covered List restriction applies; that equipment must use certification.

### Keep the compliance file

Retain at least:

- final schematic, PCB, BOM, enclosure, cables, representative supply, and photos;
- exact module grant, integration instructions, antenna evidence, and RF-exposure review;
- test configuration, procedure, equipment/calibration data, results, signatures, and setup photos;
- final firmware version and locked radio parameters;
- label/manual artwork and SDoC statement;
- production inspection procedure, model/lot traceability, and compliance-relevant changes;
- the current §2.938 Covered List production certification.

Ordinary SDoC records are generally retained for two years after marketing is permanently discontinued.

### Supply the required FCC information

Prepare this in-house after the product passes:

- a unique product/model identifier;
- a U.S.-located responsible party and contact information;
- the §15.19 two-condition Part 15 statement;
- `Contains FCC ID: 2AC7Z-ESPS3MINI1` when the module label is hidden;
- the §15.21 unauthorized-modification warning;
- the §15.105 Class B interference notice and corrective measures;
- any separation, installation, cable, supply, load, and radio restrictions required by the module and tested configuration.

The FCC logo is optional. Do not call the complete controller “FCC certified.” After the SDoC is supported, accurate wording is: **“Contains FCC ID: 2AC7Z-ESPS3MINI1; host evaluated for Part 15 Class B compliance.”**

### Other actual legal duties

- Do not offer or advertise the finished RF product for sale before its required authorization is complete; FCC “marketing” includes offering and advertising for sale.
- Track complaints, overheating, failures, and returns. A defect creating a substantial product hazard or unreasonable risk can trigger prompt CPSC reporting.
- If selling into California, implement reasonable connected-device security. Do not ship a universal default password; use secure provisioning and controlled updates.
- Do not make optional regulated claims such as `UL Listed`, `ETL Listed`, `Made in USA`, or `Wi-Fi CERTIFIED` unless the exact claim is substantiated.

No general CPSC premarket certificate or universal federal UL/ETL requirement was identified for this adult, low-voltage, batteryless controller. That changes if the product becomes a children's product, contains a button/coin cell, adds a battery, or enters another regulated category.

## 2. Work we can do ourselves

No certification body is needed for:

- module-grant and integration review;
- antenna/keepout checklist and firmware radio restrictions;
- initial RF-exposure calculation or grant comparison when the required data are clear;
- thermal soak, connector heating, current, output-short, polarity, PPTC, strain, and foreseeable-misuse checks;
- near-field, LISN, TEM-cell, and spectrum-analyzer pre-compliance scans;
- labels, manual, SDoC, security documentation, production controls, and change log;
- final SDoC measurements **only if** our facility, calibrated equipment, site validation, procedure, and report actually satisfy the FCC rules.

A normal room, backyard scan, SDR, or near-field probe is useful for finding problems. It is not automatically valid final evidence. Building a compliant home radiated-emissions facility is legally possible, but for one product it is unlikely to be cheaper than renting qualified lab time.

> **Certification-body-free is lawful. Measurement-free is not.**

## 3. Non-government commercial gate: Bluetooth SIG

Bluetooth qualification is **not an FCC law or government certification**. It is a private Bluetooth SIG membership, intellectual-property, qualification, and trademark program. Because Bluetooth LE is a required product feature, treat it as a commercial release gate rather than merely a logo suggestion. The SIG's current guidance says all Bluetooth products must be qualified; the member marketing this controller must complete qualification under its own account by the start of sale or distribution, with the qualified product name and model matching what is marketed.

Before paying, confirm the exact Espressif qualified-design lineage and the lowest-cost inherited-design qualification path in the Bluetooth Qualification Workspace. Reusing a qualified design may avoid new Bluetooth radio testing, but it does not automatically remove the product-qualification fee. Merely omitting the Bluetooth name or logo does not by itself resolve the SIG qualification, license, and patent analysis.

Current published fees effective 2026-03-01:

- Adopter membership: **$0/year**; product qualification: **$12,000**.
- Eligible small Contributing Adopter: **$3,500/year** plus **$8,000** for its first qualification that membership year, or **$11,500 total**.

For one product, the eligible small Contributing Adopter route is only $500 below the free-Adopter route. Confirm eligibility and the exact fee before treating either number as a quote.

## 4. Suggestions—not launch law for this configuration

These are risk reduction or sales-channel choices, not assumed legal gates:

- UL/ETL/NRTL listing;
- product-liability insurance;
- paid legal or regulatory review;
- an accredited laboratory rather than another competent SDoC facility;
- a separate laboratory pre-scan;
- independent thermal, electrical, flammability, and mechanical evaluation;
- RoHS declarations for a U.S.-only product without a covered display;
- Proposition 65 warning without first determining that a listed-chemical exposure requires one;
- CE, UKCA, ISED/Canada, WEEE, or other non-U.S. programs;
- a bundled power adapter, battery, or broader accessory kit.

Retailers, insurers, workplace customers, landlords, or local authorities can turn some suggestions—especially an NRTL mark—into contractual or site-specific requirements. Check the actual first sales channel before paying for them.

## 5. One test package and multiple enclosure shapes

One representative report may support a model family when every variant remains materially equivalent. Freeze:

- PCB and firmware;
- module orientation and antenna clearance;
- nonconductive enclosure material near the antenna;
- board mounting and cable routing;
- input supply specification and cable lengths;
- output connector, representative load, and operating modes.

Test the worst-case marketed configuration and keep a written equivalence rationale for the other shapes. Reevaluate metal, conductive paint/filament, reduced antenna clearance, changed board orientation, longer cables, a different supply, new converters/displays/motors/radios, or a significant PCB/firmware change. SDoC covers identical production, not an unlimited family by assertion.

## 6. Cheapest credible cost model

These are planning figures, not quotes.

### Fixed launch costs

| Item | Bare-minimum treatment | Planning cash |
|---|---|---:|
| FCC module certification | Reuse Espressif grant within its conditions | $0 new FCC fee |
| Part 15B SDoC filing/TCB | Self-issued; no filing or TCB | $0 |
| Labels, manual, SDoC, records | Prepare internally | Near $0 plus printing/time |
| RF-exposure/module review | DIY if the grant clearly covers use; otherwise professional review | $0–$1,500 estimate |
| Final Part 15B Class B measurements/report | Self-test only with a compliant facility; otherwise rent qualified lab time | $1,500–$4,000 estimate |
| Retest after a failure | Avoid with internal pre-scan | $500–$3,000 per round estimate |
| Bluetooth SIG product qualification | Private release gate; verify inherited-design route | $11,500–$12,000 current published path |
| UL/ETL | Defer unless an actual channel/AHJ/insurer requires it | $0 at launch |

A practical first-pass budget is therefore:

- **government/FCC work using an outside lab:** about **$1,500–$5,500**;
- **government filing and certification-body fees:** **$0** while the module grant remains valid and the host uses SDoC;
- **FCC work plus the current published Bluetooth qualification path:** about **$13,000–$17,500** before any failure/retest.

### Controller production cost already recorded

The current hardware record estimates:

- provisional electronic BOM: about **$5.90 each at quantity 5** and **$4.37 each at quantity 100**;
- earlier PnP planning figure at quantity 100: **$6.30–$7.10 each**;
- external connector/pigtail allowance at quantity 100: **$0.60–$1.50 each**.

That implies a provisional **$6.90–$8.60 controller subtotal at quantity 100**, before enclosure, shipping, tax, programming/test labor, fixture amortization, ESP32 top-side reflow labor/yield, and JLC extended-part/setup charges. It is not a finished-product quote.

At only 100 units, the expected FCC-plus-Bluetooth fixed cost adds roughly **$130–$175 per unit**, before the controller and enclosure. This is why a legal BLE launch is dominated by fixed qualification cost, not the PCB BOM.

## 7. Production game plan

1. **Freeze the first SKU:** controller only, one plastic enclosure family, no battery, adapter, LED rope, or additional radio.
2. **Close the hardware/firmware blockers:** fabrication review, conservative enforced current limit, ESP32-S3/GPIO18 alignment, final cable/supply/load ratings, and final antenna environment.
3. **Verify the exact module package:** archive the FCC grant, current Espressif integration exhibit, RF-exposure conditions, and Bluetooth qualified-design identifiers.
4. **Reduce avoidable unit cost:** reconcile JLC Extended Parts; obtain actual quotes for the current split assembly versus professional placement of the top-side module; include labor and yield rather than treating Carlos's time as free.
5. **Build production-representative samples:** final enclosure, cables, firmware, supply, and worst-case supported load.
6. **Pre-check internally:** emissions hot spots, thermal behavior, connector/fuse/load faults, security provisioning, and radio-setting restrictions.
7. **Request at least three scoped EMC quotes:** Part 15B Class B radiated and applicable conducted testing plus a limited host-transmitter/integration spot check in the same session.
8. **Confirm and purchase the Bluetooth qualification only after the design and qualification path are frozen.**
9. **Run final measurements, correct failures, and freeze the tested revision.**
10. **Issue the SDoC, publish the manual, apply labels, retain the compliance file, and only then offer the finished product for sale.**
11. **Add enclosure variants by documented equivalence; retest only when a change can materially affect RF, EMC, or exposure.**
