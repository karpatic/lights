Browser upload is not a working feature of this project today.

## Current Path

Firmware builds and uploads are PlatformIO/serial tasks outside the website. The existing environment is still generic ESP32:

```bash
pio run
pio run --target upload
```

The serial device path depends on the selected USB-to-UART adapter and host. Do not preserve `/dev/ttyUSB0`, `/dev/ttyUSB1`, CP2102, or COM-port logs as production instructions.

## V2 Fixture Intent

V2 is intended to use an external six-pad UART fixture carrying 5V, GND, TX, RX, EN, and BOOT/GPIO0. The expected ROM-loader sequence is: hold BOOT/GPIO0 low, pulse EN low, release EN, then release BOOT.

Fixture geometry and flashing must be regenerated and tested from the final portrait PCB. The archived fixture files are stale and must not be manufactured.
