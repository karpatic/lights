# Firmware
> Current firmware status and ESP32-S3 compatibility blockers.

- keywords: ['firmware', 'platformio', 'esp32-s3', 'ble']

The firmware in [`src/`](./src/) is useful project code, but it is not a V2 production image yet.

## Current Firmware Facts

- [`platformio.ini`](./platformio.ini) defines `[env:esp32dev]` with `board = esp32dev`.
- [`src/main.cpp`](./src/main.cpp) constructs the LED strip at startup from the default data struct.
- The startup defaults are 300 LEDs, GPIO15, and `maxCurrent = 8000` mA.
- [`src/communications.cpp`](./src/communications.cpp) parses BLE JSON updates for `pixelPin`, `ledCount`, `pixelCount`, `colorOrder`, and `maxCurrent`.
- `maxCurrent` is currently parsed and printed; it is not an electrical current limiter.
- Updating `pixelPin` over BLE does not necessarily rebind the already-constructed `Adafruit_NeoPixel` output.

## V2 Compatibility Blockers

1. Add or replace the PlatformIO environment for ESP32-S3-MINI-1-N8.
2. Make GPIO18 the fixed production LED data pin, or implement a verified safe reinitialization path.
3. Implement current estimation and limiting tied to LED count, brightness, and color output.
4. Set a conservative product current ceiling after bench validation.
5. Validate UART flashing and recovery through the six-pad EN/BOOT/GPIO0 fixture.
6. Validate BLE behavior and RF range in the final enclosure.

## Current Build Command

Use PlatformIO for firmware builds when the environment is installed:

```bash
pio run
```

Do not treat a generic ESP32 build or upload as proof that V2 hardware is ready.
