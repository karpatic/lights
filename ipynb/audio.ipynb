# Audio
> Current audio-reactive status.

- keywords: ['audio', 'meyda', 'dj']

Audio-reactive control is exploratory. The historical notes focused on Meyda/Web Audio feature extraction and an external DJ panel that can broadcast color events.

## Current Status

- There is no V2 audio hardware requirement in the current controller design.
- No microphone input is part of the V2 PCB contract.
- Audio analysis is a browser/application layer idea, not a release blocker for the compact ESP32-S3 LED controller.
- Any future audio-reactive path should send the same validated BLE command payloads as the browser controller after firmware compatibility is resolved.

The old Meyda feature notes are preserved at [`archive/notes/2026-07-30/audio.ipynb`](./archive/notes/2026-07-30/audio.ipynb).
